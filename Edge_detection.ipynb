```python
import cv2
import numpy as np
```

```python
#read image
image=cv2.imread("build.jpg")  #0-grayscale (black & white image)
```

```python
# sobel edge detection
sobelx=cv2.Sobel(image,cv2.CV_64F,1,0,ksize=3)
sobel=cv2.convertScaleAbs(sobelx)
sobely=cv2.Sobel(image,cv2.CV_64F,0,1,ksize=3)
sobel=cv2.convertScaleAbs(sobely)

#combine both direction
sobelc=cv2.addWeighted(sobelx,0.5,sobely,0.5,0)
```

```python
#display image

cv2.imshow("Original image",image)
cv2.imshow("D=Sobel X:",sobelx)
cv2.imshow("Sobel Y axis:",sobely)
cv2.imshow("Combined Image:",sobelc)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#Canny technique
canny=cv2.Canny(image,100,200)
cv2.imshow("Original image",image)
cv2.imshow("Canny Edge:",canny)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Laplacian technique

lap=cv2.Laplacian(image,cv2.CV_64F)
lap=cv2.convertScaleAbs(lap)

cv2.imshow("Original image",image)
cv2.imshow("Laplacion Edge:",lap)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#Corner detection using Harris Technique
#corner to grascale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
#CONVERT GRAY TO FLOAT 32 (float conversion)
gray = np.float32(gray)
```

```python
#apply hgrris technique

corner=cv2.cornerHarris(
    gray, #input image
    2, #neighbour size
    3, #Sobel kernel size
    0.04) #Harris detector parameter
```

```python
image[corner > 0.01 * corner.max()] = [0,0,255]
```

```python
cv2.imshow("Harris image:", image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
