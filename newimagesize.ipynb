```python
import cv2
import matplotlib.pyplot as plt
```

```python
# Load the image (correct file name)
image = cv2.imread("image1.jpg")
```

```python
# Check if image is loaded
if image is None:
    print("Error loading image!")
else:
    print("Loaded Successfully")
```

```python
# Resize image
resize = cv2.resize(image, (650, 650))
```

```python
 # Save resized image
cv2.imwrite("newimagesize.jpg", resize)
```

```python
# Convert BGR → RGB for display
img_rgb = cv2.cvtColor(resize, cv2.COLOR_BGR2RGB)
```

```python
# Show using matplotlib
plt.imshow(img_rgb)
plt.axis("off")
plt.show()
```

```python
  # Show using OpenCV
cv2.imshow("Image", resize)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
